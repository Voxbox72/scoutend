```python
import pandas as pd

# Step 1: Load all CSVs
keepers_2020 = pd.read_csv('2020_keepers.csv')
keepers_2021 = pd.read_csv('2021_keepers.csv')
keepers_2022 = pd.read_csv('2022_keepers.csv')
keepers_2023 = pd.read_csv('2023_keepers.csv')
keepers_2024 = pd.read_csv('2024_keepers.csv')

# Step 2: Concatenate them into one DataFrame
keepers_all = pd.concat([keepers_2020, keepers_2021, keepers_2022, keepers_2023, keepers_2024], ignore_index=True)

# Step 3: Check for nulls
null_counts = keepers_all.isnull().sum()

# Step 4: Print out columns with missing values
print(null_counts)
```

```python
keepers_all.fillna(0,inplace= True)
```

```python
null_counts = keepers_all.isnull().sum()
```
