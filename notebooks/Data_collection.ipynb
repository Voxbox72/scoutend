```python
from functools import reduce
import pandas as pd

def complete_stats_field_players(source):
    # Load dataframes for each stat type
    df1 = source.read_player_season_stats(stat_type="standard").reset_index()
    df2 = source.read_player_season_stats(stat_type="shooting").reset_index()
    df3 = source.read_player_season_stats(stat_type="passing").reset_index()
    df4 = source.read_player_season_stats(stat_type="passing_types").reset_index()
    df5 = source.read_player_season_stats(stat_type="goal_shot_creation").reset_index()
    df6 = source.read_player_season_stats(stat_type="defense").reset_index()
    df7 = source.read_player_season_stats(stat_type="possession").reset_index()
    df8 = source.read_player_season_stats(stat_type="playing_time").reset_index()
    dfs = [df1,df2,df3,df4,df5,df6,df7,df8]
    for i in dfs:
        i.columns = ['_'.join(col).strip() if col[1] else col[0] for col in i.columns]
    
    # Assign dataframes and suffixes
    data_frames = [
        (df1, '@standard'), 
        (df2, '@shooting'), 
        (df3, '@passing'), 
        (df4, '@passing_types'), 
        (df5, '@goal_shot_creation'), 
        (df6, '@defense'), 
        (df7, '@possession'), 
        (df8, '@playing_time'), 
    ]
    
    # Rename columns in each dataframe to prevent conflicts
    for df, suffix in data_frames:
        df.columns = [f"{col}{suffix}" if col not in ['league', 'season', 'team', 'player'] else col for col in df.columns]
    
    # Merge all dataframes on the specified index
    df_merged = reduce(lambda left, right: pd.merge(left, right, on=['league', 'season', 'team', 'player'], how='outer'), [df for df, _ in data_frames])
    df_merged.columns = [col.split('@')[0] if '@' in col else col for col in df_merged.columns]
    return df_merged
```

```python
from functools import reduce
import pandas as pd

def complete_stats_field_players(source):
    # Load dataframes for each stat type
    df1 = source.read_player_season_stats(stat_type="keeper").reset_index()
    df2 = source.read_player_season_stats(stat_type="keeper_adv").reset_index()
    
    dfs = [df1,df2]
    for i in dfs:
        i.columns = ['_'.join(col).strip() if col[1] else col[0] for col in i.columns]
    
    # Assign dataframes and suffixes
    data_frames = [
        (df1, '@keeper'), 
        (df2, '@keeper_adv'), 
    ]
    
    # Rename columns in each dataframe to prevent conflicts
    for df, suffix in data_frames:
        df.columns = [f"{col}{suffix}" if col not in ['league', 'season', 'team', 'player'] else col for col in df.columns]
    
    # Merge all dataframes on the specified index
    df_merged = reduce(lambda left, right: pd.merge(left, right, on=['league', 'season', 'team', 'player'], how='outer'), [df for df, _ in data_frames])
    df_merged.columns = [col.split('@')[0] if '@' in col else col for col in df_merged.columns]
    return df_merged
```

```python
df2020_keepers = complete_stats_field_players(fbref20)
df2021_keepers= complete_stats_field_players(fbref21)
df2022_keepers = complete_stats_field_players(fbref22)
df2023_keepers = complete_stats_field_players(fbref23)
```
